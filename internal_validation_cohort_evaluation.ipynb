```python
import pandas as pd
import joblib
```

```python
# Load test features
test_features_step_1 = pd.read_csv('data/test_top_5000_std_step_1_filtered.csv', index_col=0)
test_features_step_2 = pd.read_csv('data/test_top_5000_std_step_2_filtered.csv', index_col=0)
test_features_step_3 = pd.read_csv('data/test_top_5000_std_step_3_filtered.csv', index_col=0)

# Load test annotations
test_anno_step_1 = pd.read_csv('data/step_1_test_anno.csv', index_col=0)
test_anno_step_2 = pd.read_csv('data/step_2_test_anno.csv', index_col=0)
test_anno_step_3 = pd.read_csv('data/step_3_test_anno.csv', index_col=0)
```

```python
# Load models
lr_model_COAD = joblib.load('models/lr_model_COAD.pkl')
lr_model_MOC = joblib.load('models/lr_model_MOC.pkl')
lr_model_EOM = joblib.load('models/lr_model_EOM.pkl')
lr_model_STAD = joblib.load('models/lr_model_STAD.pkl')
pca_model_lr_COAD = joblib.load('models/pca_model_lr_COAD.pkl')
pca_model_lr_MOC = joblib.load('models/pca_model_lr_MOC.pkl')
pca_model_lr_EOM = joblib.load('models/pca_model_lr_EOM.pkl')
pca_model_lr_STAD = joblib.load('models/pca_model_lr_STAD.pkl')
pca_model_svm_COAD = joblib.load('models/pca_model_svm_COAD.pkl')
pca_model_svm_MOC = joblib.load('models/pca_model_svm_MOC.pkl')
pca_model_svm_EOM = joblib.load('models/pca_model_svm_EOM.pkl')
pca_model_svm_STAD = joblib.load('models/pca_model_svm_STAD.pkl')
svm_model_COAD = joblib.load('models/svm_model_COAD.pkl')
svm_model_MOC = joblib.load('models/svm_model_MOC.pkl')
svm_model_EOM = joblib.load('models/svm_model_EOM.pkl')
svm_model_STAD = joblib.load('models/svm_model_STAD.pkl')
```

```python
# Evaluate models
results_df = pd.DataFrame(index=test_anno_step_1['SampleName'])

# LR COAD
test_features_COAD_pca = pca_model_lr_COAD.transform(test_features_step_3)
predictions_COAD = lr_model_COAD.predict(test_features_COAD_pca)
results_df['LR_COAD'] = predictions_COAD

# LR POC
test_features_MOC_pca = pca_model_lr_MOC.transform(test_features_step_1)
predictions_MOC = lr_model_MOC.predict(test_features_MOC_pca)
results_df['LR_MOC'] = predictions_MOC

# LR EOM
test_features_EOM_pca = pca_model_lr_EOM.transform(test_features_step_2)
predictions_EOM = lr_model_EOM.predict(test_features_EOM_pca)
results_df['LR_EOM'] = predictions_EOM

# LR STAD
test_features_STAD_pca = pca_model_lr_STAD.transform(test_features_step_3)
predictions_STAD = lr_model_STAD.predict(test_features_STAD_pca)
results_df['LR_STAD'] = predictions_STAD

# SVM COAD
test_features_COAD_pca = pca_model_svm_COAD.transform(test_features_step_3)
predictions_COAD = svm_model_COAD.predict(test_features_COAD_pca)
results_df['SVM_COAD'] = predictions_COAD

# SVM POC
test_features_MOC_pca = pca_model_svm_MOC.transform(test_features_step_1)
predictions_MOC = svm_model_MOC.predict(test_features_MOC_pca)
results_df['SVM_MOC'] = predictions_MOC

# SVM EOM
test_features_EOM_pca = pca_model_svm_EOM.transform(test_features_step_2)
predictions_EOM = svm_model_EOM.predict(test_features_EOM_pca)
results_df['SVM_EOM'] = predictions_EOM

# SVM STAD
test_features_STAD_pca = pca_model_svm_STAD.transform(test_features_step_3)
predictions_STAD = svm_model_STAD.predict(test_features_STAD_pca)
results_df['SVM_STAD'] = predictions_STAD

results_df
```
